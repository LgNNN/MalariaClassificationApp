## Dataset Preprocessing

### Dataset split

#### Analyze sum of images per category

```python
import random
import os
import shutil

original_dataset = '/data/cell_images'
new_dataset = '/data/cell_images_split'

print(len(os.listdir(os.path.join(original_dataset, 'parasitized'))))
print(len(os.listdir(os.path.join(original_dataset, 'uninfected'))))

#13780 images in each category
train_images = 11024  # 80% of 13780 
val_images = 1378   # 10% of 13780
test_images = 1378   # 10% of 13780
```

#### Split dataset 80-10-10

```python
ratios = [train_images,val_images,test_images]
categories = ['parasitized', 'uninfected']
splits = ['train','val','test']

for split, ratio in zip(splits,ratios):
    for category in categories:
         os.makedirs(os.path.join(new_dataset, split, category), exist_ok=True)
         images = os.listdir(os.path.join(original_dataset, category))
         indices = random.sample(range(len(images)),ratio)
         for index in indices:
             src = (os.path.join(original_dataset,category ,images[index]))
             dst = (os.path.join(new_dataset, split, category ,images[index]))
             shutil.move(src,dst)
#remove original dataset directory since it will be empty after the execution
shutil.rmtree(original_dataset)
```

#### Verify sum of images in new split dataset

```python

print(len(os.listdir(os.path.join(new_dataset,'train', 'parasitized'))))
print(len(os.listdir(os.path.join(new_dataset,'train', 'uninfected'))))
print(len(os.listdir(os.path.join(new_dataset,'val', 'parasitized'))))
print(len(os.listdir(os.path.join(new_dataset,'val', 'uninfected'))))
print(len(os.listdir(os.path.join(new_dataset,'test', 'parasitized'))))
print(len(os.listdir(os.path.join(new_dataset,'test', 'uninfected'))))
```
